```python
import requests
```

```python
h_url = 'http://openapi.seoul.go.kr:8088/5763414a6567656e3939577755474d/json/TvEmgcHospitalInfo/1/100/'

h_response = requests.get(h_url)
```

```python
h_data = h_response.json()
h_data
```

```python
import pandas as pd
```

```python
temp = []
for d in h_data['TvEmgcHospitalInfo']['row']:

  temp.append(d)
```

```python
h_uni = pd.DataFrame(temp)
h_uni
```

```python
pdData = pd.DataFrame(h_data)
pdData
```

```python
for n, row in h_uni.iterrows():
  print(n, 'DUTYDIVNAM	', row['WGS84LON'], row['WGS84LAT'])
```

```python
import folium
```

```python
anMap = folium.Map(location=[37.4965, 126.9573],
                   zoom_start=11,
                   zoom_control=False,
                   control_scale=True)
anMap
```

```python
for n, row in h_uni.iterrows():
  mk = folium.Marker(location=[row['WGS84LON'], row['WGS84LAT']], popup=folium.Popup(row['DUTYNAME'], max_width=200))
  mk.add_to(anMap)
anMap
```

```python
from folium.plugins import MarkerCluster
mc = MarkerCluster().add_to(anMap)

for n, row in h_uni.iterrows():
  folium.Marker(location=[ row['WGS84LAT'],row['WGS84LON']],
                popup = folium.Popup(row['DUTYNAME'], max_width=200)
                ).add_to(mc)
anMap
```
